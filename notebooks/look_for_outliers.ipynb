```python
import pandas as pd
import os
import numpy as np
```

```python
data_dir = '/archive/vision/beery/movebench/studies_10ind'
```

```python
# iterate through all studies and load trajectories.csv
all_traj = []
for study in os.listdir(data_dir):
    traj_path = os.path.join(data_dir, study, 'trajectories.csv')
    if os.path.exists(traj_path):
        traj_df = pd.read_csv(traj_path)
        traj_df['study'] = study
        all_traj.append(traj_df)
```

```python
all_traj
```

```python
# count different outliers in the possible outlier columns. we need to first search for the work "outlier" in the column name then count for each study in each outlier column
outlier_counts = {}
for traj_df in all_traj:
    study = traj_df['study'].iloc[0]
    outlier_cols = [col for col in traj_df.columns if 'outlier' in col.lower()]
    # outlier_counts[study] = {} # if counting visible uncomment
    for col in outlier_cols:
        count = traj_df[col].sum()  # assuming outliers are marked as 1
        if study not in outlier_counts:
            outlier_counts[study] = {}
            outlier_counts[study]['total'] = len(traj_df)
        outlier_counts[study][col] = count
    # visible doesn't add info upon the outlier columns, I checked but leaving this here for documentation
    # if 'visible' in traj_df.columns:
    #     outlier_counts[study]['visible'] = (traj_df['visible']==True).sum() # count of visible frames
   
```

```python
# turn this into a dataframe for better visualization
outlier_counts_df = pd.DataFrame.from_dict(outlier_counts, orient='index').fillna(0)
```

### New way for the movebench raw csv files

```python
data_dir = '/archive/vision/beery/movebench/raw_csvs_v1'
os.listdir(data_dir)
```

```python
all_traj = []
for study in os.listdir(data_dir):
    csv_path = os.path.join(data_dir, study)
    if os.path.exists(csv_path):
        raw_df = pd.read_csv(csv_path)
        print(f"Study: {study}, Total rows: {len(raw_df)}") 
        raw_df['study'] = study
        all_traj.append(raw_df)
```

```python
outlier_counts = {}
for traj_df in all_traj:
    study = traj_df['study'].iloc[0]
    outlier_cols = [col for col in traj_df.columns if 'outlier' in col.lower()]
    # outlier_counts[study] = {} # if counting visible uncomment
    for col in outlier_cols:
        count = traj_df[col].sum()  # assuming outliers are marked as 1
        if study not in outlier_counts:
            outlier_counts[study] = {}
            outlier_counts[study]['total_points'] = len(traj_df)
            
        outlier_counts[study][col] = count
```

```python
outlier_counts
```

```python
outlier_counts_df = pd.DataFrame.from_dict(outlier_counts, orient='index').fillna(0)
outlier_counts_df['total_outliers'] = outlier_counts_df[[col for col in outlier_counts_df.columns if 'outlier' in col.lower()]].sum(axis=1)
if outlier_counts_df['import-marked-outlier'].sum()==0:
    print("No import-marked-outliers found, dropping the column.")
    outlier_counts_df.drop('import-marked-outlier', axis=1, inplace=True)
else:
    print(f"Import-marked-outliers found, keeping the column. {outlier_counts_df['import-marked-outlier'].sum()} outliers marked.")
```

```python
top_outliers = outlier_counts_df[outlier_counts_df['total_outliers']>0].sort_values('total_outliers', ascending=False)#.to_csv('outlier_counts_movebench_raw.csv')
```

## Label outliers according to speed/GPS:

```python
from movebench.datasets.thumbnails import load_thumbnail_source_df
carter_raw =load_thumbnail_source_df('/data/vision/beery/scratch/shir/codes/vibecleaning/data/movement_example/raw_v1/Carter_28708a9c_raw_merged.csv')
carter_raw.head()
```

```python
carter_raw['system-outlier'] =  (carter_raw['gps:hdop']==9.9) |(carter_raw['height-above-msl']>=200)# low quality GPS signal, thresholds assessed qualitatively by trying to reverse engineer the algorithm-marked-outliers
carter_raw['speed-outlier'] = (carter_raw['speed_m_s']>8)# abnormal speed, thrshold determined qualitatively by trying to reverse engineer the algorithm-marked-outliers
```

## Summary of findings:
disappointingly, it appears the the combination of gps:hdop==9.9 (there appears to be a cap in the raw data) and height-above-msl>=200 explains all of the algorithm outliers. 
Regarding the manual outliers - speed_m_s>8 captured 7 out of 10, but flagged 23 other fixes that are not labeled as outliers (and 18 that were already algorithm-marked).
It can be cut down by thresholding step_length at 300 but at the cost of >200 false positives so I decided to live with these 3 undetected manual outliers.
